```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.neighbors import NearestNeighbors
neighbors = NearestNeighbors(n_neighbors = 5)
distances, indices = neighbors.fit(X).kneighbors(X)
distances = np.sort(distances[:, 4])
plt.plot(distances)
plt.title('K-distance Graph (for choosing eps)')
plt.xlabel('Points sorted by distance')
plt.ylabel('5th nearest neighbor distance')
plt.show()
```

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps = 8, min_samples = 5)
y_dbscan = dbscan.fit_predict(X)
```

```python
y_dbscan
```

```python
supervised = dataset
```

```python
supervised
```

```python
supervised['Cluster_group'] = y_dbscan
```

```python
supervised
```

```python
supervised.to_csv('cluster_dbscan.csv', index=False)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4],
                   hue=supervised.columns[5], fit_reg=False, legend=True, legend_out=True)
```
