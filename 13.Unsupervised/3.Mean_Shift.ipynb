```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv('Mall_Customers.csv')
```

```python
dataset
```

```python
X = dataset.iloc[:, 3:5].values
```

```python
X
```

```python
from sklearn.cluster import MeanShift, estimate_bandwidth
bandwidth = estimate_bandwidth(X, quantile = 0.2)
meanshift = MeanShift(bandwidth = bandwidth)
y_meanshift = meanshift.fit_predict(X)
```

```python
centroids = meanshift.cluster_centers_
centroids
```

```python
y_meanshift
```

```python
supervised = dataset
```

```python
supervised
```

```python
supervised['Cluster_group'] = y_meanshift
```

```python
supervised
```

```python
supervised.to_csv('cluster_meanshift.csv', index=False)
```

```python
import seaborn as sns
facet = sns.lmplot(data=supervised, x=supervised.columns[3], y=supervised.columns[4],
                   hue=supervised.columns[5], fit_reg=False, legend=True, legend_out=True)
```
